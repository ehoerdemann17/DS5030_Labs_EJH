1. A random sample has

   $$
   \bar{X}=52,\qquad s=10,\qquad n=100.
   $$

   Compute the estimated standard error of $\bar{X}$. Test

   $$
   H_0:\mu=50
   \qquad \text{vs.} \qquad
   H_1:\mu\ne 50
   $$

   at significance levels $\alpha=.10,.05,.01$ using the large-sample $z$-statistic. These correspond to 90%, 95%, and 99% confidence levels. Construct a 95% confidence interval for $\mu$.

```python
# Imports
import pandas as pd
import numpy as np
import math
from scipy.stats import norm

# First Calculate the Estimated Standard error of X
# Use formula ESE_X = s/sqrt(n)
s = 10
n = 100
ESE_X = s/math.sqrt(n)
print(f"Estimated Standard error of Sample Mean of X: {ESE_X}")

# Next, Calculate the z-statistic (need for hypothesis testing)
X_sample_mean = 52
mu = 50
z = (X_sample_mean - mu)/ESE_X
print(f"Z-Statistic: {z}")

p = 2 * (1 - norm.cdf(abs(z)))
print(f"Two-Sided p-value: {p}")

# Now, test at each significance level
print()
significances = [0.01, 0.05, 0.1]
for a in significances:
    critical_val = norm.ppf(1 - a/2)
    if abs(z) > critical_val:
        dec = "Reject Null Hypothesis"
    else:
        dec = "Fail to Reject Null Hypothesis"
    print(f"Alpha: {a}, Critical Value: {critical_val}, Decision: {dec}")

# Now construct CIs
print(f"\n90% Confidence Interval: [{X_sample_mean - (ESE_X*1.645)},{X_sample_mean + (ESE_X*1.645)}]")
print(f"95% Confidence Interval: [{X_sample_mean - (ESE_X*1.96)},{X_sample_mean + (ESE_X*1.96)}]")
print(f"99% Confidence Interval: [{X_sample_mean - (ESE_X*2.576)},{X_sample_mean + (ESE_X*2.576)}]")

```

2. A health survey compares a biomarker to a benchmark value of 12. The sample has

   $$
   \bar{X}=11.5,\qquad s=3,\qquad n=36.
   $$

   Is 12 a plausible value for the population mean? Construct a 95% confidence interval and compute the two-sided p-value for $H_0:\mu=12$ versus $H_1:\mu\ne12$. Assess plausibility using the interval and the test at $\alpha=.05$.

```python
# First Calculate the Estimated Standard error of X
# Use formula ESE_X = s/sqrt(n)
s = 3
n = 36
ESE_X = s/math.sqrt(n)
print(f"Estimated Standard error of Sample Mean of X: {ESE_X}")

# Next, Calculate the z-statistic (need for hypothesis testing)
X_sample_mean = 11.5
mu = 12
z = (X_sample_mean - mu)/ESE_X
print(f"Z-Statistic: {z}")

p = 2 * (1 - norm.cdf(abs(z)))
print(f"Two-Sided p-value: {p}")

# Now, test at each significance level
print()
a = 0.05
critical_val = 1.96
if abs(z) > critical_val:
    dec = "Reject Null Hypothesis"
else:
    dec = "Fail to Reject Null Hypothesis"
print(f"Alpha: {a}, Critical Value: {critical_val}, Decision: {dec}")

# Now construct CIs
print(f"\n95% Confidence Interval: [{X_sample_mean - (ESE_X*1.96)},{X_sample_mean + (ESE_X*1.96)}]")
```

**Answer**: The interval and the test both point to the conclusion that 12 is a plausible value for the population mean. 12 resides inside of the 95% CI, and the p-value is far greater than 0.05 at the 0.05 significance level, so we fail to reject the null hypothesis that the population mean is 12.

3. A real estate analyst compares the average price of a house in two neighborhoods.
   Group A:

   $$
   \bar{X}_A=25,\qquad s_A=8,\qquad n_A=100
   $$

   Group B:

   $$
   \bar{X}_B=22,\qquad s_B=7,\qquad n_B=100.
   $$

   Compute the two-sample $Z$-statistic for testing equality of means. Construct an approximate 95% confidence interval for $\mu_A-\mu_B$.


```python
# First define variables

# Group A
X_A_mean = 25
sA = 8
nA = 100

# Group B
X_B_mean = 22
sB = 7
nB = 100

# Now find ESE_diff
ESE_diff = math.sqrt((sA**2)/nA + (sB**2)/nB)
print(f"Estimated Standard Error of Difference in Group Means: {ESE_diff}")

diff = X_A_mean - X_B_mean
z = (diff - 0)/ESE_diff
print(f"Difference of Sample Means: {diff}")
print(f"Z-Statistic: {z}")

p = 2 * (1 - norm.cdf(abs(z)))
print(f"Two-Sided p-value: {p}")

# Now, test at each significance level
print()
a = 0.05
critical_val = 1.96
if abs(z) > critical_val:
    dec = "Reject Null Hypothesis"
else:
    dec = "Fail to Reject Null Hypothesis"
print(f"Alpha: {a}, Critical Value: {critical_val}, Decision: {dec}")

# Now construct CIs
print(f"\n95% Confidence Interval: [{diff - (ESE_diff*1.96)},{diff + (ESE_diff*1.96)}]")
```

Let $\mu$ denote the population mean of this recorded-time variable, and test $H_0:\mu=120$ months. Complete the full inference routine and interpret the CIs in both months and years. 7. For `./data/nhanes_data_17_18.csv`, use `AgeInYearsAtScreening` to select respondents aged 18 or older with observed `SleepHoursWeekdaysOrWorkdays`. Estimate mean weekday sleep and test $H_0:\mu=8$ hours. Comment on whether statistical and practical significance tell the same story.
```python
# Note of AI Usage: I needed some serious help on this question, as well as the rest of the questions. I did not know alot of the syntax, and used Claude Opus 5.5 to help me write the majority of the code

# 4

# Load the data and define the indicator (1 if Central.Air == 'Y', else 0)
ames = pd.read_csv("C:/users/ericj/DS5030/DS5030_Labs_EJH/Homework/homework_data/ames_prices.csv")
x = (ames["Central.Air"] == "Y").astype(int).to_numpy()

# Point estimate: sample proportion = mean of the indicator
theta_hat = x.mean()
theta0 = 0.90   # H0: p = 0.90
print(f"n = {len(x)}")
print(f"Estimated proportion with central air: {theta_hat}")

# a. Bootstrap with a reproducible RNG
seed = 42
R = 2000
rng = np.random.default_rng(seed)
print(f"\nSeed: {seed}, R: {R}")

boot_props = np.empty(R)
for r in range(R):
    sample = rng.choice(x, size=len(x), replace=True)
    boot_props[r] = sample.mean()

# b. Bootstrap standard error (ddof=1 means denominator R - 1)
SE_boot = boot_props.std(ddof=1)
print(f"Bootstrap SE: {SE_boot}")

# d. Test statistic and p-value
z = (theta_hat - theta0) / SE_boot
p = 2 * norm.cdf(-abs(z))
print(f"\nZ-Statistic: {z}")
print(f"Two-Sided p-value: {p}")

# c, d, e. CIs, test decisions, and agreement check at each level
print()
critical_values = {0.10: 1.645, 0.05: 1.96, 0.01: 2.576}
for a, crit in critical_values.items():
    lower = theta_hat - crit * SE_boot
    upper = theta_hat + crit * SE_boot
    contains = lower <= theta0 <= upper

    if p < a:
        dec = "Reject Null Hypothesis"
    else:
        dec = "Fail to Reject Null Hypothesis"

    agrees = (dec == "Reject Null Hypothesis") == (not contains)
    print(f"{int((1-a)*100)}% CI: [{lower:.4f}, {upper:.4f}], contains 0.90: {contains}")
    print(f"  Alpha: {a}, Decision: {dec}, Agrees with CI: {agrees}")
```

**Answer for Parts e, f:** We reject H0 when p = 0.90 is outside of the CI. Since 0.90 is outside of the CI at every confidence level, we reject the null hypothesis unequivocally.


**Estimate Interpretation:** Approximately 93% of homes in the Ames sample have Central Air

**Interval:** We are 95% confident that the true proportion of Ames homes that have Central Air is between ~[0.92, 0.94]

**Test**: Since the p-value is 5.5 which is much larger than any of the test thresholds we fail to reject the null hypothesis. In other words, we cannot say with evidence that the proportion of homes with Central Air is different than 90%.
